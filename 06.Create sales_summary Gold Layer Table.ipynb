```sql
create or refresh materialized view restaurant.gold.d_sales_summary
tblproperties ("quality" = "gold")
as
select
order_date,
count(order_id) as total_orders,
cast(sum(total_amount) as decimal(10,2)) as total_revenue,
cast(sum(total_amount) / count(order_id) as decimal(10,2)) as avg_order_value,
count(distinct customer_id) as unique_customers,
count(distinct restaurant_id) as unique_restaurants,
sum(case when order_type = 'delivery' then 1 else 0 end) as delivery_orders,
sum(case when order_type = 'dine_in' then 1 else 0 end) as dine_in_orders,
sum(case when order_type = "takeaway" then 1 else 0 end) as takeaway_orders
from live.restaurant.silver.fact_orders
group by order_date
```
