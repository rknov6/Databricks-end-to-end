```python
from pyspark.sql.functions import col,to_date
```

```python
df = spark.table("restaurant.bronze.customers") \
          .withColumn("join_date",col("join_date").cast("date"))
df.writeTo("restaurant.silver.dim_customers").createOrReplace()
```

```python
df_restau = spark.table("restaurant.bronze.restaurants") \
                 .withColumn("opening_date",to_date(col("opening_date")))
df_restau.writeTo("restaurant.silver.dim_restaurants").createOrReplace()
```

```python
df_menu = spark.table("restaurant.bronze.menu_items") \
               .withColumn("is_vegetarian",col("is_vegetarian").cast("boolean"))
df_menu.writeTo("restaurant.silver.dim_menu_items").createOrReplace()
```
